#### Forward propagation

$$relu(x) = x\ if\ x > 0\ else\ 0$$

```python
def forward(X, W1, b1, W2, b2):
    Z = X @ W1 + b1
    # Z активированные нейроны
    Z_act = Z * (Z > 0) # Маска: отрицательные значения занулятся сами собой
    Y_pred = Z_act @ W2 + b2 # на веса и сдвиг
    return Z_act, Y_pred
```
